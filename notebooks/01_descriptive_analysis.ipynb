## ۹. شمردن مقدارهای گمشده در کل فایل

در سلول قبل فقط پنج نظر را بررسی کردیم. این بار `chunksize=100_000` فایل را به تکه‌هایی با حداکثر ۱۰۰٬۰۰۰ ردیف تقسیم می‌کند و حلقه همهٔ تکه‌ها را به ترتیب می‌خواند.

`missing_counts` از صفر شروع می‌شود و برای هر ستون یک شمارنده دارد. داخل حلقه، `chunk.isna().sum()` تعداد خانه‌های گمشدهٔ همان تکه را می‌دهد و آن را به شمارش قبلی اضافه می‌کنیم. هم‌زمان `len(chunk)` را جمع می‌زنیم تا تعداد ردیف‌های کل فایل را داشته باشیم.

فعلاً مقادیری را می‌شماریم که pandas در زمان خواندن CSV گمشده تشخیص می‌دهد. عدد `0.0` همچنان مقدار موجود است. تعداد ردیف‌ها شامل سطر عنوان ستون‌ها نیست.


```python
missing_counts = pd.Series(0, index=df.columns, dtype="int64")
total_rows = 0

for chunk in pd.read_csv(data_path, chunksize=100_000):
    missing_counts = missing_counts + chunk.isna().sum()
    total_rows = total_rows + len(chunk)

print(f"Total reviews: {total_rows:,}")
missing_counts
```

## ۱۰. درصد مقدارهای گمشده در کل فایل

از دو نتیجهٔ سلول ۹ استفاده می‌کنیم: `missing_counts` تعداد خانه‌های گمشده برای هر ستون و `total_rows` تعداد کل نظرهاست. برای هر ستون، تعداد گمشده را بر تعداد کل ردیف‌ها تقسیم و در ۱۰۰ ضرب می‌کنیم. فایل دوباره خوانده نمی‌شود.

`pd.DataFrame` تعداد و درصد را کنار هم در یک جدول می‌گذارد. درصد را برای نمایش تا دو رقم اعشار گرد می‌کنیم. `sort_values` ستون‌های دارای بیشترین مقدار گمشده را بالا می‌آورد.


```python
missing_percent = (missing_counts / total_rows) * 100

missing_summary = pd.DataFrame({
    "missing_count": missing_counts,
    "missing_percent": missing_percent.round(2),
})

missing_summary.sort_values("missing_count", ascending=False)
```

## ۱۱. شمارش برچسب‌های پیشنهاد محصول

در قدم قبل، خانه‌های گمشده را شمردیم. این بار می‌خواهیم بدانیم بین نظرهایی که `recommendation_status` دارند، هر برچسب چند بار تکرار شده است.

`usecols` می‌گوید فقط ستون `recommendation_status` را وارد جدول کنیم. `value_counts(dropna=True)` هر برچسب موجود را می‌شمارد و خانه‌های خالی را کنار می‌گذارد. `no_idea` خودش یک برچسب ثبت‌شده است، پس شمرده می‌شود.

با `add(..., fill_value=0)` شمارش هر برچسب را به جمع قبلی اضافه می‌کنیم. اگر برچسبی در یکی از دو طرفِ جمع دیده نشده، سهم آن طرف صفر است. در پایان شمارش‌ها را به عدد صحیح تبدیل می‌کنیم و بیشترین شمارش را اول می‌آوریم.


```python
recommendation_counts = pd.Series(dtype="int64")

for chunk in pd.read_csv(
    data_path, usecols=["recommendation_status"], chunksize=100_000
):
    chunk_counts = chunk["recommendation_status"].value_counts(dropna=True)
    recommendation_counts = recommendation_counts.add(chunk_counts, fill_value=0)

recommendation_counts = recommendation_counts.astype("int64").sort_values(ascending=False)
recommendation_counts
```

## ۱۲. سهم هر برچسب بین نظرهای برچسب‌دار

`recommendation_counts.sum()` تعداد نظرهایی را می‌دهد که وضعیت پیشنهادشان ثبت شده است. تعداد هر برچسب را بر این عدد تقسیم می‌کنیم و در ۱۰۰ ضرب می‌کنیم تا سهم آن برچسب به درصد به دست بیاید.

در سلول ۱۰ مخرج تعداد کل نظرها بود؛ اینجا مخرج فقط تعداد نظرهای برچسب‌دار است. `no_idea` یکی از برچسب‌های ثبت‌شده است؛ خانه‌های خالی در هیچ‌کدام از شمارش‌های این جدول نیستند.

جدول نهایی تعداد و درصد را کنار هم نشان می‌دهد. گرد کردن تا دو رقم اعشار فقط برای نمایش است. این جدول توزیع برچسب‌های موجود در فایل خام را نشان می‌دهد، پیش از بررسی قابل‌استفاده بودن متن هر نظر برای آموزش.


```python
labeled_rows = recommendation_counts.sum()
recommendation_percent = (recommendation_counts / labeled_rows) * 100

recommendation_summary = pd.DataFrame({
    "count": recommendation_counts,
    "percent": recommendation_percent.round(2),
})

recommendation_summary
```

## ۱۳. نمودار توزیع برچسب‌های پیشنهاد

از جدول سلول ۱۲ استفاده می‌کنیم: نام برچسب‌ها روی محور افقی و درصد نظرهای برچسب‌دار روی محور عمودی است. `ax.bar(...)` برای هر برچسب یک میله می‌سازد و `ax.bar_label(...)` درصد را بالای همان میله می‌نویسد.

منبع نمودار: کل نظرهای دارای برچسب در فایل خام؛ خانه‌های خالی `recommendation_status` در مخرج نیستند. محور عمودی از ۰ تا ۱۰۰ درصد است تا اختلاف سه کلاس در مقیاس واقعی دیده شود.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(recommendation_summary.index, recommendation_summary["percent"])
ax.bar_label(bars, fmt="%.2f%%", padding=3)

ax.set_title("Recommendation label distribution")
ax.set_xlabel("Recommendation label")
ax.set_ylabel("Share of labeled reviews (%)")
ax.set_ylim(0, 100)

fig.tight_layout()
plt.show()
```

## ۱۴. چه مقدارهایی در ستون rate ثبت شده؟

مثل شمارش برچسب‌ها در سلول ۱۱، فقط ستون `rate` را تکه‌تکه می‌خوانیم و تعداد تکرار هر مقدار را جمع می‌کنیم. این شمارش روی کل فایل خام است و هیچ ردیفی حذف نمی‌شود.

این بار `dtype="string"` مقدارهای امتیاز را در مرحلهٔ خواندن به شکل متن نگه می‌دارد. `keep_default_na=False` هم اجازه نمی‌دهد رشته‌های خالی یا متن‌هایی مثل `NA` خودکار به مقدار گمشده تبدیل شوند؛ اگر چنین مقدارهایی در فایل باشند، جداگانه در شمارش دیده می‌شوند. این دو گزینه برای مشاهدهٔ مقدارهای ثبت‌شده‌اند و به معنی معتبر دانستن همهٔ آن‌ها نیستند.

`value_counts()` تعداد هر مقدار را می‌شمارد. بعد شمارش تکه‌ها را با `add(..., fill_value=0)` جمع می‌زنیم. `sort_index()` در این مرحله مقدارهای متنی را مرتب می‌کند، نه مقدارهای عددی را.

`to_string()` شمارش همهٔ مقدارهای متفاوت را چاپ می‌کند تا ردیف‌های وسط جدول در نمایش خلاصه پنهان نشوند.

در قدم بعد، عددی بودن، محدودهٔ مجاز و معنی صفر را بررسی می‌کنیم. فعلاً به صفر هیچ معنی تازه‌ای نمی‌دهیم و فایل اصلی را تغییر نمی‌دهیم.


```python
raw_rate_counts = pd.Series(dtype="int64")

for chunk in pd.read_csv(
    data_path, usecols=["rate"], chunksize=100_000,
    dtype="string", keep_default_na=False
):
    chunk_counts = chunk["rate"].value_counts()
    raw_rate_counts = raw_rate_counts.add(chunk_counts, fill_value=0)

raw_rate_counts = raw_rate_counts.astype("int64").sort_index()
print(raw_rate_counts.to_string())
```

## ۱۵. تبدیل کنترل‌شدهٔ امتیاز و بررسی محدوده

از شمارش کامل سلول ۱۴ یک جدول می‌سازیم: `raw_rate` مقدار اصلی و `count` تعداد نظرهایی است که آن مقدار را دارند. هر ردیف این جدول یک امتیاز متفاوت است، نه یک نظر.

`pd.to_numeric(..., errors="coerce")` متن‌های عددی را به عدد تبدیل می‌کند و متن غیرقابل‌تبدیل را در ستون جدید `numeric_rate` با `NaN` نشان می‌دهد. ستون مقدار اصلی حفظ می‌شود و تعداد موارد غیرقابل‌تبدیل را صریحاً گزارش می‌کنیم؛ هیچ ردیفی خودکار حذف نمی‌شود.

`notna()` مشخص می‌کند تبدیل به عدد انجام شده است. `between(0, 5)` بررسی می‌کند عدد بین ۰ و ۵، با احتساب دو مرز، باشد. `~` نتیجهٔ شرط را برعکس می‌کند؛ `&` یعنی هر دو شرط هم‌زمان برقرار باشند.

با `loc[شرط, "count"]` شمارش مقدارهای مطابق شرط را انتخاب و جمع می‌کنیم. بنابراین اگر یک مقدار نامعتبر برای چندین نظر تکرار شده باشد، همهٔ آن نظرها در گزارش حساب می‌شوند.

صفر در این بررسی داخل محدودهٔ عددی است؛ هنوز دربارهٔ معنای آن نتیجه نمی‌گیریم. مقدار `2500.0` را هم حدسی اصلاح نمی‌کنیم. نامعتبر بودن امتیاز، به‌تنهایی مجوز حذف متن یا برچسب پیشنهادِ همان نظر نیست. این سلول فقط گزارش می‌دهد و CSV اصلی را تغییر نمی‌دهد.


```python
rate_audit = pd.DataFrame({
    "raw_rate": raw_rate_counts.index,
    "count": raw_rate_counts.values,
})

rate_audit["numeric_rate"] = pd.to_numeric(rate_audit["raw_rate"], errors="coerce")
is_numeric = rate_audit["numeric_rate"].notna()
in_range = rate_audit["numeric_rate"].between(0, 5)

non_numeric_count = rate_audit.loc[~is_numeric, "count"].sum()
out_of_range_count = rate_audit.loc[is_numeric & ~in_range, "count"].sum()
invalid_rates = rate_audit.loc[~in_range]

print("Reviews with non-numeric rates:", non_numeric_count)
print("Reviews with out-of-range rates:", out_of_range_count)
invalid_rates
```

## ۱۶. تعداد، کمترین، بیشترین و میانگین امتیازهای داخل محدوده

از شرط `in_range` در سلول ۱۵ استفاده می‌کنیم تا فقط امتیازهای عددی بین ۰ و ۵ را وارد این محاسبات کنیم. مقدار صفر داخل این محدوده است؛ هنوز دربارهٔ معنای آن فرض تازه‌ای نداریم. فایل اصلی و جدول اعتبارسنجی حفظ می‌شوند.

`valid_rates` همچنان جدول مقدارهای متفاوت و تعداد تکرار آن‌هاست. پس برای تعداد نظرها باید ستون `count` را جمع کنیم؛ تعداد ردیف‌های این جدول، تعداد نظرها نیست.

برای میانگین، هر امتیاز را در تعداد تکرارش ضرب می‌کنیم. مجموع این حاصل‌ضرب‌ها را بر تعداد نظرهای دارای امتیاز داخل محدوده تقسیم می‌کنیم. این همان میانگینی است که از تک‌تک امتیازهای این نظرها می‌گرفتیم، بدون ساختن دوبارهٔ فهرست چندمیلیونی امتیازها.

مثلاً اگر امتیاز ۱ دو بار و امتیاز ۵ یک بار ثبت شده باشد، میانگین برابر است با `(1*2 + 5*1) / 3`، نه `(1+5) / 2`.

این آمار روی کل فایل و پس از کنار گذاشتن امتیازهای نامعتبر از همین محاسبه است. `:.2f` فقط میانگین چاپ‌شده را تا دو رقم اعشار نشان می‌دهد و مقدار متغیر را گرد نمی‌کند. میانه و پراکندگی را در قدم‌های بعد بررسی می‌کنیم.


```python
valid_rates = rate_audit.loc[in_range].copy()

valid_rate_count = valid_rates["count"].sum()
min_rate = valid_rates["numeric_rate"].min()
max_rate = valid_rates["numeric_rate"].max()

weighted_rate_sum = (valid_rates["numeric_rate"] * valid_rates["count"]).sum()
mean_rate = weighted_rate_sum / valid_rate_count

print(f"Reviews with rates in [0, 5]: {valid_rate_count:,}")
print("Minimum rate:", min_rate)
print("Maximum rate:", max_rate)
print(f"Mean rate: {mean_rate:.2f}")
```

## ۱۷. میانهٔ امتیازهای داخل محدوده

میانه مقدار وسطِ امتیازهای مرتب‌شده است. اگر تعداد امتیازها زوج باشد، دو مقدار وسط را پیدا و میانگین آن‌ها را حساب می‌کنیم. اگر فرد باشد، دو جایگاه محاسبه‌شده در کد یکی می‌شوند.

`valid_rates` جدول مقدارهای متفاوت و تعداد تکرار آن‌هاست. گرفتن مستقیم `median()` از ستون `numeric_rate` تکرار امتیازها را در نظر نمی‌گیرد؛ برای همین از ستون `count` استفاده می‌کنیم.

اول با `sort_values` جدول را بر اساس امتیاز عددی مرتب می‌کنیم. سپس `cumsum()` تعداد تکرارها را به شکل تجمعی جمع می‌زند. مثلاً برای امتیازهای ۱، ۳ و ۵ با تعدادهای ۲، ۱ و ۱، شمارش تجمعی ۲، ۳ و ۴ می‌شود. یعنی جایگاه‌های ۱ و ۲ متعلق به امتیاز ۱، جایگاه ۳ متعلق به امتیاز ۳ و جایگاه ۴ متعلق به امتیاز ۵ است.

جایگاه‌ها را از ۱ می‌شماریم. `//` تقسیم صحیح است. دو فرمول `(N+1)//2` و `(N+2)//2` هم برای تعداد زوج و هم فرد کار می‌کنند. برای N=4، دو جایگاه وسط ۲ و ۳ هستند؛ برای N=5 هر دو فرمول جایگاه ۳ را می‌دهند.

برای هر جایگاه، اولین ردیفی را می‌گیریم که شمارش تجمعی‌اش به آن جایگاه رسیده یا از آن گذشته است. `iloc[0]` یعنی اولین ردیفِ نتیجهٔ فیلتر. در مثال بالا، دو امتیاز وسط ۱ و ۳ و میانه برابر ۲ است.

این محاسبه از شمارش کاملِ امتیازهای داخل محدودهٔ سلول ۱۶ استفاده می‌کند و نیازی به خواندن دوبارهٔ فایل یا ساختن فهرست چندمیلیونی ندارد.


```python
sorted_rates = valid_rates.sort_values("numeric_rate").copy()
sorted_rates["cumulative_count"] = sorted_rates["count"].cumsum()

left_position = (valid_rate_count + 1) // 2
right_position = (valid_rate_count + 2) // 2

left_rate = sorted_rates.loc[
    sorted_rates["cumulative_count"] >= left_position, "numeric_rate"
].iloc[0]
right_rate = sorted_rates.loc[
    sorted_rates["cumulative_count"] >= right_position, "numeric_rate"
].iloc[0]

median_rate = (left_rate + right_rate) / 2

print(f"Middle positions (counting from 1): {left_position:,}, {right_position:,}")
print("Middle rates:", left_rate, right_rate)
print("Median rate:", median_rate)
```

## ۱۸. واریانس و انحراف معیار امتیازها

میانگین یک مرکز برای امتیازهاست. انحراف معیار نشان می‌دهد امتیازها چقدر حول این مرکز پراکنده‌اند.

اول اختلاف هر امتیاز با `mean_rate` را حساب می‌کنیم و اختلاف‌ها را به توان ۲ می‌رسانیم. توان ۲ باعث می‌شود اختلاف‌های مثبت و منفی در جمع همدیگر را خنثی نکنند. بعد هر اختلافِ مربع‌شده را در تعداد تکرار همان امتیاز ضرب می‌کنیم.

مجموع را بر تعداد امتیازها تقسیم می‌کنیم تا واریانس به دست بیاید. جذر واریانس، انحراف معیار است. در پایتون `** 2` یعنی توان ۲ و `** 0.5` یعنی جذر.

در این گزارش تمام امتیازهای داخل محدودهٔ فایل را توصیف می‌کنیم و تعریف جامعه‌ای را با تقسیم بر N به کار می‌بریم (`ddof=0`). این با پیش‌فرض `pandas.std()` که برای انحراف معیار نمونه‌ای بر N-1 تقسیم می‌کند، متفاوت است.

از مقدار دقیق `mean_rate` در حافظه استفاده می‌کنیم، نه میانگین ۳٫۶۵ که برای نمایش گرد شده بود. انحراف معیار در واحد خودِ امتیاز است؛ واریانس در واحد مربع امتیاز. این محاسبه دربارهٔ پراکندگی امتیازهای ثبت‌شده است، نه خطای پیش‌بینی یک مدل.

داده همان امتیازهای داخل محدودهٔ ۰ تا ۵ از کل فایل است؛ امتیاز صفر همچنان در محاسبه حضور دارد. جدول‌های قبلی تغییر نمی‌کنند و CSV دوباره خوانده نمی‌شود.


```python
deviations = valid_rates["numeric_rate"] - mean_rate
squared_deviations = deviations ** 2

weighted_squared_sum = (squared_deviations * valid_rates["count"]).sum()
variance_rate = weighted_squared_sum / valid_rate_count
std_rate = variance_rate ** 0.5

print(f"Population variance (ddof=0): {variance_rate:.4f}")
print(f"Population standard deviation (ddof=0): {std_rate:.4f}")
```

## ۱۹. چارک‌های امتیازهای داخل محدوده

Q1، Q2 و Q3 به‌ترتیب صدک‌های ۲۵، ۵۰ و ۷۵ هستند. آن‌ها جایگاه‌های یک‌چهارم، وسط و سه‌چهارمِ امتیازهای مرتب‌شده را توصیف می‌کنند. Q2 همان میانه است. به دلیل تکرار امتیازها، این مرزها الزاماً چهار گروه با تعداد دقیقاً برابر ایجاد نمی‌کنند.

از `sorted_rates` و ستون `cumulative_count` سلول ۱۷ استفاده می‌کنیم تا تکرار هر امتیاز در محاسبه حساب شود. هیچ امتیازی از فایل دوباره خوانده نمی‌شود.

قرارداد محاسبه، درون‌یابی خطی است؛ همان `quantile(..., interpolation="linear")` در pandas. برای N امتیاز و نسبت q، جایگاه یک‌مبنا برابر `1 + (N - 1) * q` است. مثلاً برای `[1, 1, 3, 5]` و q=0.75، جایگاه ۳٫۲۵ می‌شود: یک‌چهارمِ راه بین امتیاز جایگاه سوم (۳) و چهارم (۵). بنابراین Q3 برابر ۳٫۵ است.

`math.floor` و `math.ceil` جایگاه پایین و بالا را پیدا می‌کنند. مثل سلول ۱۷، اولین شمارش تجمعی که به هر جایگاه رسیده یا از آن گذشته باشد، امتیاز آن جایگاه را می‌دهد. `fraction` می‌گوید چه سهمی از فاصلهٔ بین دو امتیاز را حرکت کنیم. اگر جایگاه صحیح باشد، هر دو امتیاز یکی‌اند.

حلقه این مراحل را برای سه نسبت ۰٫۲۵، ۰٫۵۰ و ۰٫۷۵ تکرار می‌کند. نتیجه فقط دربارهٔ امتیازهای عددی داخل محدودهٔ ۰ تا ۵ در کل فایل است، با همان سیاست سلول‌های قبل.


```python
import math

quartiles = {}

for label, q in {"Q1 (25%)": 0.25, "Q2 (50%)": 0.50, "Q3 (75%)": 0.75}.items():
    position = 1 + (valid_rate_count - 1) * q
    lower_position = math.floor(position)
    upper_position = math.ceil(position)

    lower_rate = sorted_rates.loc[
        sorted_rates["cumulative_count"] >= lower_position, "numeric_rate"
    ].iloc[0]
    upper_rate = sorted_rates.loc[
        sorted_rates["cumulative_count"] >= upper_position, "numeric_rate"
    ].iloc[0]

    fraction = position - lower_position
    quartiles[label] = lower_rate + (upper_rate - lower_rate) * fraction

quartile_summary = pd.Series(quartiles, name="rate")
quartile_summary
```
